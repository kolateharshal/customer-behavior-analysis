```python
df = sns.load_dataset("titanic") # Reload the original dataset

print(df.columns.tolist())

#Keep only the columns we actually want to use as features + the target.

df = df[["survived", "pclass", "sex", "age", "sibsp", "parch", "fare", "embarked"]].copy()

df.head()
```

```python
print(df.isnull().sum())
```

```python
df["age"] = df["age"].fillna(df["age"].median())

#"embarked" (port) has just 2 missing values -> fill with the MODE #(the most common port) since it's categorical, not numeric.

df["embarked"]= df["embarked"].fillna (df["embarked"].mode()[0])

#Confirm nothing is missing anymore.

print(df.isnull().sum())
```

```python
df["age"] = df["age"].fillna(df["age"].median())

#"embarked" (port) has just 2 missing values -> fill with the MODE

#(the most common port) since it's categorical, not numeric.

df["embarked"] = df["embarked"].fillna(df["embarked"].mode()[0])

# Confirm nothing is missing anymore.

print(df.isnull().sum())
```

```python
df["family_size"] = df["sibsp"] + df["parch"] + 1 #+1 to count the passenger

#Encode "sex" as 0/1 (this is a simple manual encoding fine for a binary cati

df["sex"] = df["sex"].map({"male": 0, "female": 1})

#Encode "embarked" (3 possible ports) as 0/1/2.

df["embarked"]= df["embarked"].map({"S": 0, "C": 1, "Q": 2})

df.head()
```

```python
X = df[["pclass", "sex", "age", "fare", "family_size", "embarked"]]

y = df["survived"]

print("Features used:", X.columns.tolist())

print("Target: survived (e did not survive, 1 survived)")
```

```python
RANDOM_STATE = 42
X_train, X_test, y_train, y_test = train_test_split( X, y, test_size=0.2, random_state=RANDOM_STATE )

print("Training rows:", X_train.shape[0])

print("Testing rows:", X_test.shape[0])
```
